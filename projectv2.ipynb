# Cleaning "Customer Call List.xlsx"

Importing the necessary libraries

```python
import pandas as pd
```

Getting the data from a given Excel file

```python
df = pd.read_excel("Customer Call List.xlsx")
```

Preview of the Data

```python
df.head()
```

Dropping the duplicates

```python
df = df.drop_duplicates()
```

Dropping non-useful columns

```python
df.drop(columns = "Not_Useful_Column",inplace = True)
```

Fixing the column "Last_Name"

```python
df["Last_Name"] = df["Last_Name"].str.strip("1._/")
```

Fixing the column "Phone_Number"

```python
df["Phone_Number"] = df["Phone_Number"].str.replace("-","")
df["Phone_Number"] = df["Phone_Number"].str.replace("|","")
df["Phone_Number"] = df["Phone_Number"].str.replace("/","")
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : str(x))
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : x[0:3] + "-" + x[3:6] + "-" + x[6:10])
```

Splitting the "Address" column into three more intuitive columns

```python
df[["Street_Name","State","Zip"]] = df["Address"].str.split(",",n = 2,expand = True)
df["State"] = df["State"].str.strip()
df["Zip"] = df["Zip"].str.strip()
```

Standardising columns "Paying Customer" And "Do_Not_Contact"

```python
df["Paying Customer"] = df["Paying Customer"].replace("Yes","Y")
df["Paying Customer"] = df["Paying Customer"].replace("No","N")
df["Do_Not_Contact"] = df["Do_Not_Contact"].replace("Yes","Y")
df["Do_Not_Contact"] = df["Do_Not_Contact"].replace("No","N")
```

Marking blank "Do_Not_Contact" values as "Y", then removing all Do Not Contact customers

```python
df["Do_Not_Contact"] = df["Do_Not_Contact"].fillna("Y")
df = df[df["Do_Not_Contact"] != "Y"]
```

Dropping rows with null value in "Phone_Number"

```python
df["Phone_Number"] = df["Phone_Number"].replace(["nan--", "Na--"], "")
df = df[df["Phone_Number"]!= ""]
```

Resetting the Index

```python
df = df.reset_index(drop = True)
df.head()
```

Getting the cleaned data in a CSV file

```python
df.to_csv("cleaned_customer_list.csv", index=False)
```

# Changes Made:
